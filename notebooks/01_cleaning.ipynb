```python
# validate="one_to_one" will raise an error instead of blindly merging if
# either side has duplicates
# Merging on="id" keeps a single id column, so no extra deletion required
df = f1.merge(f2, on="id", how="inner", validate="one_to_one")
print("Merged_shape:", df.shape)
```

```python
# Finding hidden missing values

# 1. True nulls: cells that are really empty
null_counts = df.isna().sum()
print("True Nulls:")
print(null_counts[null_counts > 0])

# 2. Fake nulls: text that means missing
# astype(str) turns every value into text so we can compare them all,
# strip() removes spaces, upper() makes "n/a" and "N/A" look the same
suspects = ["", "NA", "N/A", "NULL", "NONE", "NAN", "-"]

# NOTE: astype(str) turns a real NaN into the text "nan", so "NAN" below
# is the true nulls counted a second time, not new missing values
print("\nFake nulls:")
for col in df.columns:
    if df[col].dtype == "object":
        values = df[col].astype(str).str.strip().str.upper()
        found = values[values.isin(suspects)]
        if len(found) > 0:
            print(col, found.value_counts().to_dict())

# 3. Mixed types: what kind of values are inside each column
print("\nMixed types:")
check_cols = ["mths_since_last_delinq", "mths_since_last_record", "revol_util",
            "last_pymnt_d", "next_pymnt_d", "last_credit_pull_d"]
for col in check_cols:
    print(col)
    print(df[col].apply(type).value_counts())
```

Conclusions and decisions

39717 rows and 48 columns.

Note: "NAN" in the fake-null output is not new missing data. astype(str) turns a real NaN into the text "nan", so the true nulls show up a second time. I count them once.

1. mths_since_last_delinq ("NA": 25,682) [Meaning - borrower was never late on a payment] - set to null + add flag ever_delinquent

2. mths_since_last_record ("NA": 36,931) [Meaning - no public record, e.g. bankruptcy] - set to null + add flag has_public_record

3. next_pymnt_d (38,577 blanks, 1,140 dates) [Meaning - 1,140 matches the number of Current loans, so maybe only active loans have a next payment date] - keep as null, then verify the link to Current loans

4. last_pymnt_d (71 blanks) [Meaning - probably the loan never received a payment] - keep as null, then verify

5. revol_util (50 blanks) [Unknown] - keep as null (not 0), then check against revol_bal

6. last_credit_pull_d (2 blanks) [Unknown] - keep as null and keep the rows. Deleting these 2 whole loans along with their money would make 39,715 loans; a blank in a column that isn't important should not quietly change every financial total. Correct loan count is 39,717.

7. desc (13,153 in total: 12,939 empty + 210 space-only + 3 "NONE" + 1 "N/A") [Borrower didn't write a description] - set to null + add flag has_desc

8. emp_title (2,453 empty + 6 "N/A" = 2,459), emp_length (1,075 "N/A"), title (9 empty + 4 text = 13) [Borrower didn't provide the information] - set to null

9. home_ownership ("NONE": 3) [It's a real category] - leave as is.

```python
# Convert fake missing values to real ones
import numpy as np
# for each column: which text values mean missing?
# separate list per column, because "NONE" is real data in home_ownership

fake_nulls = {
    "emp_title": ["", "N/A"],
    "emp_length": ["N/A"],
    "desc": ["", "NONE", "N/A"],
    "title": ["", "NONE", "N/A", "NA"],
    "mths_since_last_delinq": ["NA"],
    "mths_since_last_record": ["NA"],
    "revol_util": [""],
    "last_pymnt_d": [""],
    "next_pymnt_d": [""],
    "last_credit_pull_d": [""],
}

for col in fake_nulls:
    # same cleanup as the detection step, so we match exactly what we found there
    clean_text = df[col].astype(str).str.strip().str.upper()
    is_fake = clean_text.isin(fake_nulls[col])
    df.loc[is_fake, col] = np.nan
    print(col, "->", is_fake.sum(), "converted to NaN")

# Flags: keep the meaning of the blank as a yes(1)/no(0) column
df["ever_delinquent"]   = df["mths_since_last_delinq"].notna().astype(int)
df["has_public_record"] = df["mths_since_last_record"].notna().astype(int)
df["has_desc"]          = df["desc"].notna().astype(int)
```

astype(str) creates a temporary text copy that we use only to find the fake nulls. The real column is changed only in the rows flagged in is_fake, so the real dates and numbers are never touched.

So basically, we normalized the text the same way we did during detection, built a True/False mask of fake-null values per column, and set only those cells to NaN.

Counts: only the text values get converted here (emp_title 6, emp_length 1,075, desc 214, title 4, and the two months-since columns). revol_util and the three date columns show 0 because their blanks were already NaN when loading. They stay in the dictionary as a safety net. After this step, the missing totals match the conclusions above exactly.

```python
# Fix data types
# term: "36 months" -> 36
df["term"] = df["term"].str.replace("months", "").str.strip().astype(int)

# int_rate: "11.89%" -> 0.1189 (fraction, same scale as revol_util)
df["int_rate"] = (df["int_rate"].str.replace("%", "").astype(float) / 100).round(4)

# emp_length: "< 1 year" -> 0, "10+ years" -> 10, missing stays missing
# extract(r"(\d+)") simply means "take the first number in the text"
df["emp_length_yrs"] = (df["emp_length"]
                        .str.replace("< 1", "0")
                        .str.extract(r"(\d+)", expand=False)
                        .astype(float))

# issue_d: "Aug-09" -> real date (%b = short month name, %y = 2-digit year)
df["issue_d"] = pd.to_datetime(df["issue_d"].str.strip(), format="%b-%y")

# Month counts: Int64 = whole numbers that ALLOW blanks
df["mths_since_last_delinq"] = pd.to_numeric(df["mths_since_last_delinq"]).astype("Int64")
df["mths_since_last_record"] = pd.to_numeric(df["mths_since_last_record"]).astype("Int64")

df["revol_util"] = pd.to_numeric(df["revol_util"])

df["last_pymnt_d"]       = pd.to_datetime(df["last_pymnt_d"])
df["next_pymnt_d"]       = pd.to_datetime(df["next_pymnt_d"])
df["last_credit_pull_d"] = pd.to_datetime(df["last_credit_pull_d"])

print(df.dtypes)

# Check the emp_length conversion: every text value next to its number
print(df[["emp_length", "emp_length_yrs"]].drop_duplicates())
```

```python
# Prove that each blank means what we think it means
# How to read a crosstab: rows = question A (False/True), columns = question B (0/1).
# Each number = how many borrowers have that combination.
# Look at the "impossible" cell - if it is 0, the assumption is proven.
# 1. Anyone late in the last 2 years should have ever_delinquent = 1
print(pd.crosstab(df["delinq_2yrs"] > 0, df["ever_delinquent"]))

# 2. Anyone with a public record should have has_public_record = 1
print(pd.crosstab(df["pub_rec"] > 0, df["has_public_record"]))

# 3. Is 0 a REAL value? (if yes, filling blanks with 0 is wrong)
print("Real zeros:", (df["mths_since_last_delinq"] == 0).sum())

# 4. Do only Current loans have a next payment date?
print(pd.crosstab(df["loan_status"], df["next_pymnt_d"].notna()))

# 5. The 71 loans with no last payment date: what status, how much paid?
no_pay = df[df["last_pymnt_d"].isna()]
print(no_pay["loan_status"].value_counts())
print("Total paid by these 71:", no_pay["total_pymnt"].sum())

# 6. The 50 missing revol_util: what is their revolving balance?
print(df[df["revol_util"].isna()]["revol_bal"].describe())
```

What Step 7 proved

1. Delinquency flag is confirmed. The impossible cell (late recently = True, but blank = 0) is 0. All 4,312 borrowers who were late in the last 2 years have ever_delinquent = 1, with no contradictions. Another 9,723 were late at some point, but more than 2 years ago.

2. Public record flag mostly checks out, with one quirk. All 2,116 borrowers with pub_rec > 0 are flagged correctly. But 670 borrowers have a "months since record" value while pub_rec = 0. The most likely explanation is that the record existed but no longer counts, for example because it was removed or aged off. This is a quirk in the source data, not a mistake in cleaning. We keep it as it is. We don't "fix" source data just to make it look tidy.

3. 443 real zeros. 443 borrowers were late this month, which is recorded as 0 months ago. 

4. next_pymnt_d exists only for the 1,140 Current loans. Confirmed exactly. It only tells "this loan is still active," which loan_status already says. Decision: drop the column.

5. The 71 loans with no payment date are all Charged Off, but together they show $43,857 in total_pymnt. My hypothesis is that this money came from debt collection (recoveries) after the loan defaulted, not from the borrower paying. That would mean "no last payment date" is correct. Next step checks this.

6. revol_util blanks: at least 75% of those 50 borrowers have a $0 revolving balance. With no credit line to use, utilization can't be calculated (it's 0 divided by 0), so it isn't 0%. A blank is the honest value. The few with a balance are simply unknown. Keeping them as NaN is correct.

Corrected headline numbers: Charged Off = 5,627 and default rate on completed loans = 5,627 ÷ (5,627 + 32,950) = 14.59%. We exclude Current loans because we don't know yet how they'll end.

Cleaning isn't only about blanks. A value can be filled in and still be impossible, like a loan repaid before it was issued. The + 1 and > $1 allow for small rounding differences so we only catch real problems. 

The earliest_cr_line range check looks for a classic two-digit-year bug, where "62" gets read as 2062 instead of 1962.

```python
# Business-rule checks (only checks, nothing changes)
# Each line counts the rows that break a rule that must be true for real loans. We want to see 0 everywhere.
# Money rules
print("funded > requested:", (df["funded_amnt"] > df["loan_amnt"]).sum())
print("investor part > funded:", (df["funded_amnt_inv"] > df["funded_amnt"]).sum())
print("principal repaid > funded:", (df["total_rec_prncp"] > df["funded_amnt"] + 1).sum())

# total_pymnt should equal the sum of its parts
parts = df["total_rec_prncp"] + df["total_rec_int"] + df["total_rec_late_fee"] + df["recoveries"]
print("total_pymnt not matching its parts (> $1 off):", ((df["total_pymnt"] - parts).abs() > 1).sum())

# Date rules
print("credit line opened AFTER loan issued:", (df["earliest_cr_line"] > df["issue_d"]).sum())
print("earliest_cr_line range:", df["earliest_cr_line"].min(), "to", df["earliest_cr_line"].max())
print("last payment BEFORE loan issued:", (df["last_pymnt_d"] < df["issue_d"]).sum())

# Status rule: finished loans should owe nothing, Current loans should owe something
print(df.groupby("loan_status")["out_prncp"].agg(["min", "max"]))

# Ranges: look for impossible or extreme values
print(df[["int_rate", "dti", "revol_util", "annual_inc", "installment"]].describe().T)

# Test the hypothesis about the 71 loans
no_pay = df[df["last_pymnt_d"].isna()]
print("Recoveries of the 71:", no_pay["recoveries"].sum())
```

Part 1 looks for hidden variants like "Verified" vs "verified " that would split one category into two in a chart. It also finds constant columns, meaning a column with only 1 unique value. pymnt_plan and initial_list_status are likely candidates. A column that's identical for every loan carries no information, so it gets dropped.

Part 2 only touches rows that actually have a job title (has_title), so the blanks stay real NaN instead of turning into the text "nan". Lowercase instead of title case: title case would change "IBM" to "Ibm" and "US Army" to "Us Army".

```python
# Check the categories and tidy the job titles

# 1. Every category column: are the values clean and consistent?
cat_cols = ["grade", "sub_grade", "home_ownership", "verification_status",
            "loan_status", "pymnt_plan", "initial_list_status", "purpose", "addr_state"]
for col in cat_cols:
    print(col, "->", df[col].nunique(), "unique:", df[col].unique())

# 2. emp_title: "Walmart", "walmart ", "WALMART" should count as one employer
print("emp_title unique BEFORE:", df["emp_title"].nunique())
has_title = df["emp_title"].notna()
df.loc[has_title, "emp_title"] = df.loc[has_title, "emp_title"].astype(str).str.strip().str.lower()
print("emp_title unique AFTER:", df["emp_title"].nunique())
print(df["emp_title"].value_counts().head(15))
```

Conclusions

emp_title: stripping spaces and lowercasing merged 1,611 duplicate job titles (e.g. "Walmart", "walmart ", "WALMART" became one), reducing distinct titles from 28,820 to 27,209.
Every business rule passed with 0 violations. Money, dates, and statuses are all internally consistent, and the earliest_cr_line range (1946–2008) shows no two-digit-year bug.

The 71-loan mystery is solved. Their recoveries total $43,857.56, exactly equal to their total_pymnt. So these borrowers never made a single payment, and all of that money came from debt collection after the loan defaulted. Leaving their last payment date blank is 100% correct.

total_pymnt includes collection recoveries, so "Total Payment Received" isn't pure borrower repayment. The collection agency also charged fees (collection_recovery_fee). We'll use this when building a correct profit measure in the SQL step.

column, range, verdict

1. int_rate | 5.42 % - 24.59 % | Normal for grades A-G

2. dti | 0 – 29.99 |Just under the platform's 30 cap, so real

3. annual_inc | $4K – $6M | Extreme outlier, and the mean ($69K) is well above the median ($59K)

Decision on the $6M income: we will keep it. It's unusual but possible, and deleting real data because it's inconvenient is bad practice. Instead we deal with it at analysis time: we will use median income, not average, in the dashboards, because a few very high values pull the average up.

The categories are clean, with no spelling or spacing variants. There are two constant columns: pymnt_plan is always "n" and initial_list_status is always "f". They carry no information, so both get dropped.

Job titles still contain synonyms ("us army" / "usaf" / "us air force", "self" / "self employed"). Fully standardizing them would take manual mapping for no payoff, since no KPI uses job title. Decision: leave them.

```python
# Drop the justified columns, export, and verify
# Drop columns, each with a documented reason
drop_cols = ["next_pymnt_d",          # only exists for Current loans; loan_status already says this
             "pymnt_plan",            # constant: always "n"
             "initial_list_status",   # constant: always "f"
             "desc"]                  # long free text with line breaks; has_desc keeps the useful part
df = df.drop(columns=drop_cols)

print("Final shape:", df.shape)   # expected (39717, 48)

# Export
df.to_csv("../data/processed/loans_clean.csv", index=False)
```
